# Exploring silver prices and daily returns

Same idea as the gold notebook: load silver_futures.csv, look at the raw price, then compute and plot daily returns. See notebooks/01_explore_returns.ipynb for the fuller explanations of each step.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
silver = pd.read_csv("../data/silver_futures.csv", skiprows=[1, 2], index_col=0, parse_dates=True)
silver.index.name = "Date"
silver.head()
```

## Raw closing price

```python
silver["Close"].plot(title="Silver Futures Closing Price", figsize=(10, 4))
plt.ylabel("USD per troy ounce")
plt.show()
```

## Daily returns

Same formula as before: (today's close - yesterday's close) / yesterday's close.

```python
silver["Return"] = silver["Close"].pct_change()
silver[["Close", "Return"]].head()
```

## Plot the returns

Worth comparing this chart to gold's: silver is historically the more volatile of the two - the swings here are typically bigger.

```python
silver["Return"].plot(title="Silver Daily Returns", figsize=(10, 4))
plt.ylabel("Daily % change")
plt.show()
```
